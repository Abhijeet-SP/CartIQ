```python
from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
```

```python
import pandas as pd 
orders = pd.read_csv("/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/processed/master_orders.csv")
```

```python
transactions = (orders.groupby("order_id")["product_name"].apply(list).tolist())
print(type(transactions))
print(transactions[0])
```

```python
encoded = te.fit(transactions).transform(transactions)
basket = pd.DataFrame(
    encoded,
    columns=te.columns_
)
```

```python
from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(
    basket,
    min_support=0.01,
    use_colnames=True
)
```

```python
frequent_itemsets
```

```python
print(type(frequent_itemsets))
```

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.2
)

rules
```
