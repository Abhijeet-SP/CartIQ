```python
import pandas as pd
```

```python
fact_order_timing = pd.read_csv("/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/raw/fact_orders_timing.csv")
fact_order_basket_prior = pd.read_csv("/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/raw/fact_orders_basket_prior.csv")
fact_order_basket_train = pd.read_csv("/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/raw/fact_orders_basket_train.csv")
dim_products = pd.read_csv("/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/raw/dim_products.csv")
dim_departments = pd.read_csv("/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/raw/dim_departments.csv")
```

```python
print(fact_order_timing.columns)
print(fact_order_basket_prior.columns)
print(fact_order_basket_train.columns)
print(dim_departments.columns)
print(dim_products.columns)
```

### Product Recommendation Layer 1 

Input: Current Cart + User History + Context (time_of_the_day, add_to_cart_order)

```python
d1 = fact_order_timing[fact_order_timing["eval_set"] == "prior"]
d2 = fact_order_timing[fact_order_timing["eval_set"] == "train"]
```

```python
d1 = d1.merge(fact_order_basket_prior, on="order_id", how='left') 
d2 = d2.merge(fact_order_basket_train, on="order_id", how='left') 
print(d1.columns)
print(d2.columns)
```

```python
d1 = d1.merge(dim_products, on='product_id', how='left')
d2 = d2.merge(dim_products, on='product_id', how='left')
print(d1.columns)
print(d2.columns)
```

```python
d1 = d1.merge(dim_departments, on='department_id', how='left')
d2 = d2.merge(dim_departments, on='department_id', how='left')
print(d1.columns)
print(d2.columns)
```

```python
output_path_d1 = "/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/processed/order_product_data_prior.csv"
output_path_d2 = "/Users/abhijeetsinghparihar/Desktop/Projects/Supply Chain Project/CartIQ/data/processed/order_product_data_train.csv"
d1.to_csv(output_path_d1, index=False)
d2.to_csv(output_path_d2, index=False)
```
